```python
import pandas as pd

customers = pd.read_csv("../data/raw/customers.csv")
products = pd.read_csv("../data/raw/products.csv")
orders = pd.read_csv("../data/raw/orders.csv")

print("Customers:", customers.shape)
print("Products:", products.shape)
print("Orders:", orders.shape)
```

```python
print("Customers missing values:")
print(customers.isnull().sum())

print("\nProducts missing values:")
print(products.isnull().sum())

print("\nOrders missing values:")
print(orders.isnull().sum())
```

```python
print("Customer duplicates:", customers.duplicated().sum())
print("Product duplicates:", products.duplicated().sum())
print("Order duplicates:", orders.duplicated().sum())
```

```python
customers["signup_date"] = pd.to_datetime(customers["signup_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])
```

```python
import os

# Create processed data folder if it doesn't exist
os.makedirs("../data/processed", exist_ok=True)

# Save cleaned datasets
customers.to_csv("../data/processed/customers_cleaned.csv", index=False)
products.to_csv("../data/processed/products_cleaned.csv", index=False)
orders.to_csv("../data/processed/orders_cleaned.csv", index=False)

print("Cleaned datasets saved successfully!")
```

```python
print("Duplicate customer IDs:", customers["customer_id"].duplicated().sum())
print("Missing customer IDs:", customers["customer_id"].isna().sum())
```

```python
print("Duplicate product IDs:", products["product_id"].duplicated().sum())
print("Missing product IDs:", products["product_id"].isna().sum())
```

```python
print("Duplicate order IDs:", orders["order_id"].duplicated().sum())
print("Missing order IDs:", orders["order_id"].isna().sum())
```

```python
invalid_customers = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("Invalid customer references:", len(invalid_customers))
```

```python
invalid_products = orders[
    ~orders["product_id"].isin(products["product_id"])
]

print("Invalid product references:", len(invalid_products))
```

```python
print("Invalid order dates:", orders["order_date"].isna().sum())
print("Invalid signup dates:", customers["signup_date"].isna().sum())
```

```python
print("========== DATA VALIDATION SUMMARY ==========")

print("Customers:", len(customers))
print("Products:", len(products))
print("Orders:", len(orders))

print("Duplicate customer IDs:", customers["customer_id"].duplicated().sum())
print("Duplicate product IDs:", products["product_id"].duplicated().sum())
print("Duplicate order IDs:", orders["order_id"].duplicated().sum())

print("Invalid customer references:", len(invalid_customers))
print("Invalid product references:", len(invalid_products))

print("=============================================")
```
